```python
# imports

import os
from dotenv import load_dotenv
# from scraper import *
from IPython.display import Markdown, display
from huggingface_hub import login, InferenceClient
from tqdm.notebook import tqdm
from openai import OpenAI
from transformers import pipeline, AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig, TextStreamer, TextIteratorStreamer
from threading import Thread
import torch
import gc
import gradio as gr
```

```python
# Check for GPU / CPU
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("CUDA:", torch.version.cuda)

device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
# Define model ID
model_id = "meta-llama/Llama-3.2-3B-Instruct"
```

### Load Model

```python
# Load model
# This client sends requests to a remote inference provider.
client = InferenceClient(
    api_key=hf_token,
    provider="auto"
)

# Load model directly
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Load model directly onto available hardware
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map=device,
    low_cpu_mem_usage=True,
    torch_dtype=torch.float16,
)

print("Model loaded successfully!")
print("Memory footprint:", round(model.get_memory_footprint() / 1024**3, 2), "GB")
```

```python
# Define system message
system_prompt = """You are Movieverse, an intelligent, knowledgeable, and engaging AI-powered movie assistant. 
Your expertise covers the world of cinema, including Hollywood, Bollywood, Tollywood, Kollywood, Mollywood, 
Korean cinema, Japanese cinema, Chinese cinema, European cinema, and films from every other region. 
Your knowledge spans movies from the earliest days of cinema to the latest releases."""
```

```python
def to_text(content):
    if isinstance(content, str):
        return content
    if isinstance(content, dict):
        return content.get("text", "")
    if isinstance(content, list):
        return "".join(to_text(p) for p in content)
    return str(content)

def model_call(prompt, history):

    # Defining history & Prompt
    history = [{"role":h["role"], "content":to_text(h["content"])} for h in history]
    print(type(history))
    messages = [
        {"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": prompt}]

    # Streamer
    streamer = TextIteratorStreamer(
        tokenizer,
        skip_prompt=True,
        skip_special_tokens=True
        )
    
    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    ).to(device)

    # Generation parameters
    generation_kwargs = {
            **inputs,
            "max_new_tokens": 250,
            "streamer": streamer
        }

    # Run model generation in a separate thread
    thread = Thread(
        target=model.generate,
        kwargs=generation_kwargs
    )

    thread.start()
    
    # Extract Response
    result = ""
    for new_text in streamer:
        result += new_text
        yield result
    
    thread.join()
```

### MovieVerse

```python
gr.ChatInterface(fn=model_call, 
                 title="🤖 Welcome to MovieVerse" 
                 ).\
                launch(theme=gr.themes.Soft(primary_hue="indigo", secondary_hue="purple"))
```
